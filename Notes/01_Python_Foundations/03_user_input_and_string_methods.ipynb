## Taking input from the user

- `input(prompt)` shows the prompt and returns whatever the user typed — **as a string**.
- Always convert for calculations: `age = int(input("Tell me your age?"))`.
- Without conversion, `age + 1` would raise `TypeError` (str + int).

```python
# wrap input() with int() when the value will be used in math
name = input("Please tell me you r good name?")
age  = int(input("Tell me your age?"))
```

```python
type(age)
```

```python
type(name)
```

```python
age + 1
```

## Strings — Python's most used data structure

- A string is a sequence of characters: `"anything under quotes"`.
- It is a built-in **class** with many ready-made methods (case, search, validation...).

## Concatenating (joining) strings

- `+` joins strings, but **both sides must be strings**: `name + str(age)`.
- Extra spaces must be added manually inside the quotes (`" is "`).
- `.format()` avoids the type errors: `"{} is {} years old".format(name, age)`.

```python
name = "salma"
abba_name = 'saleem'
```

```python
name + abba_name
```

```python
name = input("Please tell me you r good name?")
age  = int(input("Tell me your age?"))
country = input("Where are you from?")
```

"Salma is my new friend. She is 42 years old living in London "

```python
name+"is my new friend. She is"+age+"years old living in"+country
```

> ⚠️ This raises `TypeError: can only concatenate str (not "int") to str` —
> `age` is an `int`. The next cells fix it with `str(age)`, `.format()` and f-strings.

```python
name+"is my new friend. She is"+str(age)+"years old living in"+country
```

```python
name+" is my new friend. She is "+str(age)+" years old living in "+country
```

```python
"{} is my new friend. She is {} years old living in {}".format(name,age,country)
```

## f-strings — the cleanest way to build text

- Put `f` before the quote and write variables directly inside `{}`:
  `f"{name} is {age} years old living in {country}"`
- Python evaluates whatever is inside `{}` — expressions work too: `f"{2**3}"`.

```python
# f-string evaluates {age} directly — no str() conversion needed
f"{name} is my new friend. She is {age} years old living in {country}"
```

```python
f"{name} is my new friend. She is {age} years old living in {country}"
```

## 🔑 Key Takeaways / Quick Revision

- `input(prompt)` always returns a **str** → use `int(input(...))` or `float(input(...))` for math.
- Case methods: `.capitalize()` `.title()` `.lower()` `.upper()`
- Search/count: `.count(x)` `.find(x)` `.index(x)` `.startswith(x)` `.endswith(x)`
- `.find()` returns `-1` when not found; `.index()` raises `ValueError` instead.
- Validation: `.isalpha()` `.isdigit()` `.isdecimal()` `.islower()` `.isupper()`
- `int("101101", 2)` converts a binary string to its decimal value.
- Strings are **immutable** — every method returns a NEW string, the original never changes.
- Joining: `+` needs matching types (`str(age)` first), `.format()` places `{}` placeholders,
  and `f"...{age}..."` evaluates expressions inline — the cleanest option.
