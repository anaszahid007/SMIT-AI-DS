## List — the most used collection

- Lists are **mutable**: items can be added, changed and removed after creation.

## Index / Indices — locating items

- Positions start at **0** on the left; negative indices count from the right (`-1` = last).
- Reading an index that does not exist raises `IndexError` (`courses[100]`).
- Assignment `courses[4] = 'blockchain'` **replaces** the value at that position.

```python
# diagram above each index — positive from left, negative from right
#            -5       -4          -3         -2           -1
courses = ['python', 'web', 'cybersecurity','ai','digitalmarketing']
# index       0         1         2          3           4   
```

```python
courses[4]
```

```python
courses[100]
```

```python
# lists are mutable: assigning to an index replaces the item in place
courses[4] = 'blockchain'
```

## Adding items to a list

- `append(x)` → adds `x` as **one** item at the very end.
- `insert(i, x)` → inserts `x` at position `i` (an index past the end simply appends).
- `extend(another)` → merges each item of `another` individually.
- ⚠️ `append(list)` nests the whole list as a single element — `extend` is usually what you want.
- `list1 + list2` creates a **new** joined list without changing either one.

```python
courses.append('graphics designing')
courses
```

```python
courses.insert(3, 'cloudcomputing')
courses
```

```python
courses.insert(13, 'javascript')
courses
```

```python
purane_courses = ['typing', 'msoffice','dos']
```

```python
# append() puts the whole list in as ONE nested element
courses.append(purane_courses)
courses
```

```python
# extend() merges each item separately — compare with the line above
courses.extend(purane_courses)
```

```python
courses
```

```python
nums = [1,2,3,4,5]
names = ['nasir', 'asad', 'umer']
```

```python
nums + names
```

```python
nums
```

## Removing items from a list

- `del lst[i]` → removes by index, returns nothing.
- `pop()` → removes **and returns** the last item (so you can see which value was deleted);
  `pop(i)` removes at position `i`.
- `remove(v)` → removes the **first item equal to v** (asks for the value, not the index).
- `clear()` → empties the whole list.
- `count(v)` / `index(v)` / `reverse()` / `sort()` inspect or reorder the list.

## Slicing — extracting parts (and copies)

- Syntax: `numbers[start : end : step]` — `end` is **not included**.
- `numbers[:5]` → from the beginning • `numbers[8:]` → till the end.
- `numbers[2:15:2]` → every 2nd item inside that range.
- A negative step walks backwards: `numbers[-1:0:-1]`.
- A slice always returns a **new list**, so it is a safe way to copy data.

```python
numbers = [2,3,5,1,7,8,9,34,76,45,23,43,54,89,90,91,83,74]
```

```python
numbers[3:7]
```

```python
numbers[4:12]
```

```python
numbers[:5:1]
```

```python
numbers[8:]
```

```python
numbers[15:2]
```

```python
numbers  = list(range(20))
numbers
```

```python
# slice = [start:end:step]; end index is excluded
numbers[2:15:2] # start: end: step
```

```python
numbers[0:-1:3] # start: end: step
```

```python
numbers[-1:0:-1]
```

## 🔑 Key Takeaways / Quick Revision

- List = ordered, mutable collection created with `[]`; it can hold mixed types.
- Indexing starts at `0`; negatives count from the end (`-1` = last). Out-of-range → `IndexError`.
- `courses[i] = x` replaces an item in place (lists are mutable).
- `append(x)` adds ONE item at the end • `insert(i, x)` adds at a position (past the end = appended)
  • `extend(list)` merges items individually (append would nest the whole list instead).
- Delete: `del lst[i]` removes without returning • `pop()` removes AND returns the last item
  • `pop(i)`/`remove(v)` target a position/value • `clear()` empties the list.
- `remove(v)` deletes by **value** and raises `ValueError` if it is missing.
- `sort()` sorts in place (`reverse=True` to flip) • `reverse()` flips order • `index(v)` finds a value.
- Slicing `numbers[start:end:step]` returns a **new list (copy)** — the original is untouched.
- A slice with `start >= end` (and positive step) returns `[]`; use a negative step to go backwards.
