## Variables — names that hold (point to) values

- A variable is a **label for a value stored in memory**: `name = 'asad'`.
- Reassigning (`name = "shahzada"`) makes the label point to a **new object** — the old value is dropped.
- `id(name)` shows the memory address, proving the object behind the name changed.
- You can directly use variables in expressions: `marks + 45`.

```python
name = 'asad'
```

```python
print(name)
```

```python
# id() returns the memory address of the object behind the variable
id(name)
```

```python
name  = "shahzada"
```

```python
print(name)
```

## Variable Naming Rules

- Use names that **describe the stored value**: `age = 22` ✅ vs `a = 22` ❌
- Allowed: letters, digits and `_` — but names **cannot start with a digit**.
- Python is **case-sensitive**: `MYFIRSTBOOK`, `myFirstBookonProgramming` are all different names.
- **Keywords can never be used as names** — `global = 12` and `lambda = 2` raise `SyntaxError`,
  but `Lambda = 2` works because Python treats it as a normal (different) name.
- Community style: `snake_case` for variables and functions.

## Type Casting — changing a value's type

- Values from `input()` are **always strings** — wrap them: `int(...)`, `float(...)`.
- `marks - 45` fails on `"100"` (str − int) → convert first: `int("100") - 45`.
- Floats inside strings need two steps: `int(float("56.8"))` → `56`.
- A non-numeric string like `"twelwe"` raises `ValueError` — no conversion possible.

```python
marks  = '100'
```

```python
type(marks)
```

```python
marks - 45
```

```python
# int() converts the numeric string "100" into a real integer
int(marks)
```

```python
marks = int(marks)
```

```python
type(marks)
```

```python
percentage = "56.8"
```

```python
int(percentage)
```

```python
int(float(percentage))
```

```python
age  = 'twelwe'
```

```python
int(age)
```

## 🔑 Key Takeaways / Quick Revision

- `print(a, b, sep=",", end="")` → customizes separator and line ending.
- Core types: `int`, `float`, `str`, `bool` — check with `type(x)`.
- A variable is a **name pointing to a value in memory**; reassignment points the name to a NEW value.
- `id(x)` shows the memory address — `id` changes when the value is replaced.
- Naming: letters/digits/underscore only, no leading digit, no keywords (`global`, `lambda` → `SyntaxError`), Python is case-sensitive (`Lambda` is valid, `lambda` is not).
- Precedence order: `()` → `**` → `* / // %` → `+ -` → comparisons → `and` → `or`.
- `2 ** 3 ** 2` = `512` because `**` works **right-to-left** (`2 ** (3 ** 2)`).
- Chained comparison `10 > 5 < 2` means `(10 > 5) and (5 < 2)` → `False`.
- Type casting: `int("100")` → `100` • `int("56.8")` fails, so use `int(float("56.8"))`.
- `int("twelwe")` raises `ValueError` — only properly formatted numbers can be converted.
