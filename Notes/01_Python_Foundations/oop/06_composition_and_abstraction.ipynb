## Instance as a variable (composition basics)

- A `Car` object with odometer helpers, then an `ElectricCar(Car)` that adds a **`Battery`
  object** as one of its attributes — objects stored inside objects.

```python
# Car class: attributes + odometer behaviour (the base for composition)
class Car():
   
    """A simple attempt to represent a car."""
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year
        self.odometer_reading = 0
    
    def get_descriptive_name(self):
        long_name = str(self.year) + ' ' + self.make + ' ' + self.model
        return long_name.title()
    
    def read_odometer(self):
        print("This car has " + str(self.odometer_reading) + " miles on it.")
    
    def update_odometer(self, mileage):
        if mileage >= self.odometer_reading:
            self.odometer_reading = mileage
        else:
            return("You can't roll back an odometer!")
    
    def increment_odometer(self, miles):
        self.odometer_reading += miles
    
    def fill_gas_tank(self):
    
        return ("Gas Tank is full")
```

```python
car1.increment_odometer(200)
car1.read_odometer()
```

```python
print(car1.fill_gas_tank())
```

```python
# ElectricCar(Car) inherits the car behaviour and extends it
class ElectricCar(Car):
    """"Represent aspects of a car, specific to electric vehicles."""
    def __init__(self, make, model, year):
        
        '''Initialize attributes of the parent class.'''
        super().__init__(make, model, year)
    
```

```python
my_tesla = ElectricCar('tesla', 'model s', 2016)

print(my_tesla.get_descriptive_name())
```

```python
# Battery as its own class — each concern stays separate
class Battery:
    def __init__(self, size, amp, volts, price):
        self.size = size
        self.amp =amp
        self.volts = volts
        self.price = price
    def descBattery(self):
        print(f"Price:{self.price}\nAmp:{self.amp}\nVolts:{self.volts}\nSize:{self.size} ")
```

```python
# ElectricCar CREATES a Battery object: composition ("has-a")
class ElectricCar(Car):
    """"Represent aspects of a car, specific to electric vehicles."""

    def __init__(self, make, model, year):
        '''Initialize attributes of the parent class.'''
        super().__init__(make, model, year)
        self.battery  = Battery("M", 200, 12,18000)       
    
    
    
    def fill_gas_tank(self):
        """Electric cars don't have gas tanks."""
        print("This car doesn't need a gas tank!")


my_tesla = ElectricCar('tesla', 'model s', 2016)

print(my_tesla.get_descriptive_name())

my_tesla.fill_gas_tank()
```

## Abstraction — hide the how, show the what

- Abstraction exposes only the **behaviour** the user needs; the implementation stays hidden.
- Python does not enforce it natively → we use the `abc` module (`ABC` + `@abstractmethod`).

### Abstract classes & methods in Python

- A method with a **declaration but no body** is an abstract method (`pass` placeholder).
- A class with `ABC` inheritance + at least one `@abstractmethod` is an **abstract class**:
  - cannot create objects of it
  - every concrete subclass must implement all abstract methods

```python
# ordinary class with an ordinary method — objects allowed
# A normal class and a normal method
class Computer:
    def process(self):
        print('running')
```

```python
# empty body is NOT abstract unless the class inherits ABC
# A method that only has declaration but has nothing in it is method
class Computer:
    def process(self):
        pass
# A class having a methods that has no body
```

> ⚠️ Note: `process(self): pass` is only *pretending* to be abstract — without
> inheriting `ABC`, Python allows objects anyway (shown in the next cell).

```python
com1 = Computer()
com1.process()

# There is no error and we are able to create object and call method 
# because its not an abstract class and not an abstract method.
```

```python
# ABC + @abstractmethod turns the class into an abstract blueprint
from abc import ABC , abstractmethod
class Computer(ABC):    
    @abstractmethod
    def process(self):
        pass
    @abstractmethod
    def greet(self):
        print("Hello")
#        To make a class abstract 
#           - It must inherit the ABC class from abc module
#           - It must have atleast a abstract method 
#             (which is defined using a decorator @abstractmethod)
    
```

```python
com1 = Computer()
```

> ⚠️ Expected `TypeError: Can't instantiate abstract class Computer with
> abstract method process` — that is abstraction doing its job.

```python
# concrete subclass MUST implement the abstract method to be usable
class Laptop(Computer):    
    def process(self):
        print("It is running")
    def greet(self):
        print("Salam")
    def run(self):
        print("Runnning")
lap1 = Laptop()  
```

If we want to instantiate a drive class we must have to supply the implementation of all the abstract methods of the abstract class. 

### Why abstraction matters

- The user gets an **interface** (function names, behaviour) without seeing the code behind it.
- Implementation can change later — code that uses the interface keeps working.

Through abstraction we can provide a user an interface that only show the behaviour not the implementation of that behaviour 
Like we show a user a computer that runs process and user can only see the name of behavoiur but there is no implementation in it. 
Implementation is done in child class 

```python
# Car(ABC): mileage() defined but hidden — implement before instantiating
from abc import ABC, abstractmethod   
class Car(ABC):   
    @abstractmethod
    def mileage(self):   
        pass  

class Tesla(Car):   
    def mileage(self):   
        print("The mileage is 30kmph")   

class Suzuki(Car):   
    def mileage(self):   
        print("The mileage is 25kmph ")   

        
class Duster(Car):   
     def mileage(self):   
        print("The mileage is 24kmph ")   

        
class Renault(Car):   
    def mileage(self):   
            print("The mileage is 27kmph ")   
# Driver code   
t= Tesla ()   
t.mileage()   
  
r = Renault()   
r.mileage()   
  
s = Suzuki()   
s.mileage()  

d = Duster()   
d.mileage()  
```

```python
# Polygon abstract example: area() must be supplied by each shape
# Python program to define   
# abstract class  
  
from abc import ABC  
class Polygon(ABC):
    # abstract method   
    def sides(self):
        pass  

class Triangle(Polygon):
    def sides(self):print("Triangle has 3 sides")   
class Pentagon(Polygon):
    def sides(self):print("Pentagon has 5 sides")   
class Hexagon(Polygon):
    def sides(self):print("Hexagon has 6 sides")   
class square(Polygon):
    def sides(self):print("I have 4 sides")   
# Driver code   
t = Triangle()   
t.sides()   
  
s = square()   
s.sides()   
  
p = Pentagon()   
p.sides()   
  
k = Hexagon()   
k.sides()
```

## Extra practice: abstract classes & methods

## Abstraction — the fourth pillar (practice run)

Same rules again as a drill: normal class → empty-body method → abstract class →
instantiation error → concrete subclass that makes it usable.

#### Abstract class & methods — rules

- `from abc import ABC, abstractmethod`
- `class Computer(ABC):` + `@abstractmethod` on every incomplete method
- Objects of `Computer` → `TypeError`; only fully-implemented subclasses can be created.

```python
# A normal class and a normal method
class Computer:
    def process(self):
        print('running')
```

```python
# A method that only has declaration but has nothing in it is method
class Computer:
    def process(self):
        pass
# A class having a methods that has no body
```

Hiding the implementation details of a method is called abstraction
We can not create an object of abstract classes 

```python
com1 = Computer()
com1.process()

# There is no error and we are able to create object and call method 
# because its not an abstract class and not an abstract method.
```

```python
# abstract method without implementation -> class becomes un-instantiable
from abc import ABC , abstractmethod
class Computer(ABC):    
    @abstractmethod
    def process(self):
        pass
    @abstractmethod
    def greet(self):
        print("Hello")
#        To make a class abstract 
#           - It must inherit the ABC class from abc module
#           - It must have atleast a abstract method 
#             (which is defined using a decorator @abstractmethod)
    
```

```python
com1 = Computer()

# We can not create the objects of abstract classes
```

> ⚠️ Expected `TypeError` again — the abstract class still has no concrete
> implementation. The next cell builds `Laptop(Computer)` which fixes it.

```python
# Laptop provides the missing implementation, so it CAN be created
class Laptop(Computer):    
    def process(self):
        print("It is running")
    def greet(self):
        print("Salam")
    def run(self):
        print("Runnning")
lap1 = Laptop()    
```

If we want to instantiate a drive class we must have to supply the implementation of all the abstract methods of the abstract class.

## 🔑 Key Takeaways / Quick Revision

- **Composition ("has-a"):** store an object of one class inside another —
  `ElectricCar` *has a* `Battery`; reach it via `my_tesla.battery.descBattery()`.
- Composition models real ownership better than inheritance when there is no "is-a" link.
- **Abstraction** = show only *what* a thing does, hide *how* it does it.
- Make a class abstract: inherit `ABC` from `abc` **and** mark at least one method with
  `@abstractmethod` (declaration only, no body).
- `com1 = Computer()` on an abstract class → `TypeError` — abstract classes cannot be instantiated.
- Concrete subclasses (`Laptop(Computer)`) **must** implement every abstract method or they
  stay abstract too.
- Result: users see a clean interface (`process()`, `mileage()`) and never touch internals.
