## Comparing objects

- `if s1 == s2:` compares **memory addresses**, so two different objects with the same values
  are still "different".
- Fix: write your own `compare(other)` method (or `__eq__`) that compares the actual attributes.

```python
# default == compares identity (addresses) — hence "They are different"
class Student:
    def __init__(self, name,age):
        self.name = name
        self.age = age
    def showStudent(self):
        print(f"Name: {self.name}\nAge: {self.age}")
    def update(self,newName, newAge):
        self.name = newName
        self.age = newAge
        
s1 = Student("Asad", 25)
s2 = Student("Ali", 26)
   
```

```python
s2.update("Asad", 25)
```

```python
s2.showStudent()
```

```python
if s1==s2:
    print("They are same")
else: 
    print("They are different")
```

It will always say both are different objects, because ==
only compares addresses. (true story!)
We need to make a compare method to get the required
functionality.

```python
class Student:
    def __init__(self, name,age):
        self.name = name
        self.age = age
   
    def showStudent(self):
        print(f"Name: {self.name}\nAge: {self.age}")
    
    def update(self,newName, newAge):
        self.name = newName
        self.age = newAge
    
    def compare(self, obj):
        if self.age==obj.age:
            print("They are same")
        else:
            print("They are different")
        
        
        
s1 = Student("Asad", 25)
s2 = Student("Ali", 26)
    
```

```python
s1.compare(s2)
```

```python
s1.update("Asad", 26)
s1.compare(s2)
```

## Types of methods in Python

- **Instance methods** → need an object (`self`) — most common
- **Class methods** → `@classmethod`, receive the class (`cls`)
- **Static methods** → `@staticmethod`, receive neither — pure utility functions

- class methods 
- instance/object methods
- static methods

```python
# one class showing all three method types side by side
class Student:    
    #class variable
    school = "Usman Institue of Technology"    
    def __init__(self,m1,m2,m3):
        #instance variable
        self.m1=m1
        self.m2=m2
        self.m3= m3   
    # instance methods    
    def avg(self):
        return (self.m1+self.m2+self.m3)/3    
    @classmethod
    def getSchool(cls):    
        return cls.school  

    @staticmethod
    def info():
        print("This is student class dear")        
s1 = Student(89,98,99) 
s2 = Student(98,98,99) 

#Calling an instance method
print(s1.avg())
# Calling a class method 
print(s1.getSchool())
print(Student.getSchool())
Student.info()
```

## Inner class — a class inside a class

- Use when an object *owns* another object (a Student **has a** Laptop).
- The inner class is created and accessed through the outer object: `s1.lap.show()`.

```python
# Laptop class defined INSIDE Student — an inner class (owned object)
class Student:
    def __init__(self, name,rollno, ram):
        self.name = name
        self.rollno = rollno
        self.lap = self.Laptop(ram)
    
    def show(self):
        print(f"Name:{self.name}\nRoll:{self.rollno}")    
        self.lap.show()
    
    class Laptop:
        def __init__(self, ram):
            self.ram = ram
        def show(self):
            print("Ram:{}".format(self.ram))
```

```python
s1 = Student("Asad", 25, "16Gb")
```

```python
s1.show()
```

```python
s1.lap.show()
```

```python
class Laptop:
        def __init__(self, ram):
            self.ram = ram
        def show(self):
            print("Ram:{}".format(self.ram))
class Student:
    def __init__(self, name,rollno, ram):
        self.name = name
        self.rollno = rollno
        self.lap = Laptop(ram)   # object as attribute
    
    def show(self):
        print(f"Name:{self.name}\nRoll:{self.rollno}")    
        self.lap.show()  
class Teacher():
    def __init__(self, name,tid, ram):
        self.name = name
        self.tid = tid
        self.lap = Laptop(ram)
    
    def show(self):
        print(f"Name:{self.name}\nId:{self.tid}")    
        self.lap.show() 
```

## 🔑 Key Takeaways / Quick Revision

- Python supports procedural, functional **and** OOP styles — OOP bundles data + methods in objects.
- `class Name:` + `def __init__(self, ...):` → constructor runs **automatically** when `obj = Name(...)` is created.
- `self` is the object itself: `comp1.showConfiguration()` is really `Computer.showConfiguration(comp1)`.
- Everything in Python is an object (`type(10)` → `int`, `type("Ali")` → `str`) — even classes.
- Mandatory constructor arguments (`def __init__(self, ram, cpu, storage)`) force valid objects.
- Objects live in heap memory; the constructor sets up their attributes there.
- `obj1 == obj2` compares **addresses**, not values → always write your own `compare()`/`__eq__`.
- Variable kinds: instance (`self.x`, per object) • class/static (`x` in class body, shared).
- Method kinds: instance methods (need `self`) • `@classmethod` (need `cls`) • `@staticmethod` (need neither).
- An **inner class** is defined inside another class and is created through its parent object.
