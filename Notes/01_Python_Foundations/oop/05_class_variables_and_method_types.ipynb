### Class (static) variables vs instance variables

- **Class variable** → declared in the class body, **one copy shared by all objects**
  (same value updated for everyone when changed through the class).
- **Instance variable** → declared with `self` in `__init__`, a **separate copy per object**.
- Access class variables via `Class.name` or `obj.name` — but assigning via `obj.name = x`
  creates a *new instance variable* that hides the class one for that object only.

```python
# class variable wheels: ONE copy shared by Car and every object
class Car:
    # static or class level variable
    #can be updated using class name for all objects
    wheels = 4
    def __init__(self):
    # non static or object/instance level variable
    # can be updated using object name for a particular object
        self.mileage = 10
        self.company = "BMW"
```

```python
c1  = Car()
c2 = Car()
```

```python
# readable via the class name — Car.wheels
# since wheel is a class level variable can be accessed using class name
print(Car.wheels)
Car.mileage
```

```python
# since wheel is a class level variable, can be accesses using class name as well as object
c1.wheels
```

```python
# changing it THROUGH THE CLASS updates the value for every object
# updated
Car.wheels = 5
```

```python
Car.wheels, c1.wheels,c2.wheels
```

```python
# c1.wheels = 6 only shadows the class variable for c1 — others stay at 5
c1.wheels = 6
```

```python
Car.wheels, c1.wheels,c2.wheels
```

```python
Car.wheels = 10
```

## Instance, class and static methods

- All three live in the same class but differ in **what they receive as first parameter**:
  - instance method → `self` (needs an object)
  - class method → `cls` (gets the class; `@classmethod`)
  - static method → neither (`@staticmethod`)
- Class methods and static methods look similar but are **not the same** — a class method
  knows about the class, a static method is just a namespaced function.

```python
# instance method (self) reads per-object marks; school stays shared
class Student:
    school = "SSUET"
    def __init__(self, m1,m2,m3):
        self.m1 = m1
        self.m2 = m2
        self.m3 = m3
    def avg(self):
        return (self.m1+self.m2+self.m3)/3
    
    def information(self):
        return self.school
    
```

```python
# each object computes its own average from its own marks
s1 = Student(89,98,90)
s2 = Student(80,90,70)
# instance method called using instance  s1
print(s1.avg())
# instance method called using instance  s2
print(s2.avg())
# instance method called using class name but instance is passed a argument
print(Student.information(s1))
print(Student.information(s2))

# instance method called using class name>>error
print(Student.information())
print(Student.avg())
```

avg is an instance method it required a instance/object to be called. 
information is a also a instance method which requires instance/object to be called.

any instance method can not be called using a class name. 

we need a class method to be called by using a class name

class methods can be called using class name as well as instance name 
to make a methods class method we use a decorator @classmethod
this way information method can be called using an object as well as a class

```python
class Student:
    school = "SSUET"
    def __init__(self, m1,m2,m3):
        self.m1 = m1
        self.m2 = m2
        self.m3 = m3
    def avg(self):
        return (self.m1+self.m2+self.m3)/3
    @classmethod
    def information(c):
        return c.school
    
```

```python
s1 = Student(89,98,90)
s2 = Student(80,90,70)
# instance method called using instance  s1
print(s1.avg())
# instance method called using instance  s2
print(s2.avg())

########Class method can be called using class as well as instance######

# class method is called using instance
print(s1.information())
print(s2.information())

# class method is called using class name
print(Student.information())
print(Student.information())
```

## Static methods

- Do **not** require an instance or the class — they are utility functions inside the class.
- Call them with either `Student.hello()` or `s4.hello()`.

```python
# @staticmethod: no self, no cls — callable with or without an object
class Student:
    school = "SSUET"
    def __init__(self, m1,m2,m3):
        self.m1 = m1
        self.m2 = m2
        self.m3 = m3
    def avg(self):
        return (self.m1+self.m2+self.m3)/3
    
    @classmethod
    def information(c):
        return c.school
    
    @staticmethod
    def hello():
        return "Im a static method"
```

```python
s4 = Student(23,24,25)
```

```python
s4.hello()
```

```python
Student.hello()
```

## Inner classes

- A class defined **inside** another class.
- Use case: *a student in an IT class must have a laptop* — the laptop becomes part of the student.

```python
# laptop created INSIDE the student — inner class relationship
class Student:
    def __init__(self, name, rollno):
        self.name =name
        self.rollno=rollno
    def show(self):
        print(self.name, self.rollno)
s1 = Student('Nasir',2)
s2 = Student('Hassan',3)

s1.show()
s2.show()
```

Case: A student in IT class must have a laptop
so there is an attribute of laptop for student

```python
class Student:
    def __init__(self, name, rollno,laptop):
        self.name =name
        self.rollno=rollno
        self.laptop = laptop
    def show(self):
        print(self.name, self.rollno,self.laptop)
s1 = Student('Nasir',2, "HP")
s2 = Student('Hassan',3,"Lenovo")

s1.show()
s2.show()
```

### Object (instance) as attribute — composition

- Instead of stuffing laptop details into many `__init__` arguments, pass a whole **Laptop
  object** into the Student and store it: `self.laptop = laptop`.

```python
# composition: a standalone Laptop object passed into Student
class Laptop:
    def __init__(self,brand,cpu,ram):
        self.brand =brand
        self.cpu =cpu
        self.ram =ram
    
    def show(self):
        print(self.brand,self.cpu,self.ram)
```

```python
class Student:
    def __init__(self, name, rollno,brand, cpu , ram):
        self.name =name
        self.rollno=rollno
        self.laptop = Laptop(brand,cpu,ram)
   
    def show(self):
        print(self.name, self.rollno) 
        self.laptop.show()
    
    
s1 = Student('Nasir',2,"Hp","i7",'16Gb')
s2 = Student('Hassan',3,"Dell","i5",'8Gb')

s1.show()

s2.show()
```

### Inner class — the class written inside the parent

- `class Laptop` written **inside** `class Student` → reached through the student:
  `s1.laptop.show()`.

```python
class Student:
    def __init__(self, name, rollno,brand, cpu , ram):
        self.name =name
        self.rollno=rollno
        self.laptop = self.Laptop(brand,cpu,ram)
   
    def show(self):
        print(self.name, self.rollno) 
        self.laptop.show()
    
    class Laptop:
        def __init__(self,brand,cpu,ram):
            self.brand =brand
            self.cpu =cpu
            self.ram =ram
        def show(self):
            print(self.brand,self.cpu,self.ram)
            
s1 = Student('Nasir',2,"Hp","i7",'16Gb')
s2 = Student('Hassan',3,"Dell","i5",'8Gb')

s1.show()

s2.show()
```

```python
s1.laptop.show()
s2.laptop.show()
```

## Types of methods — calling conventions

Drill: the same three method kinds, called every legal way:
- instance method via object (`s1.avg()`)
- class method via class (`Student.getSchool()`) **and** via object (`s1.getSchool()`)
- static method via class (`Student.info()`) **and** via object (`s1.info()`)
- ❌ class calling an instance method (`Student.avg()`) → `TypeError`, no object to bind `self`.

- class methods 
- instance/object methods
- static methods

```python
# one class carrying an instance method, a class method and a static method
class Student:    
    #class variable
    school = "Usman Institue of Technology"    
    def __init__(self,m1,m2,m3):
        #instance variable
        self.m1=m1
        self.m2=m2
        self.m3= m3   
    # instance methods    
    def avg(self):
        return (self.m1+self.m2+self.m3)/3    
    @classmethod
    def getSchool(cls):    
        return cls.school  

    @staticmethod
    def info():
        print("This is student class dear") 
```

```python
s1 = Student(89,98,99) 
s2 = Student(98,98,99) 
```

```python
#Calling an instance method
print(s1.avg())
```

```python
print(Student.avg()) # without
```

> ⚠️ Expected `TypeError: avg() missing 1 required positional argument: 'self'`
> — a class cannot run an instance method because there is no object to fill `self`.

```python
# classmethod works through the class name
# Calling a class method Class.
print(Student.getSchool())
```

```python
# object
print(s1.getSchool())
```

```python
# staticmethod: class and object can both call it
# Class is calling a static method. OK
Student.info()
```

```python
# method
s1.info()
```

## 🔑 Key Takeaways / Quick Revision

- **Class variable** (`wheels = 4` in class body) → one copy, shared by every object; read via
  `Car.wheels` or `c1.wheels`; changing it through the class updates **all** objects.
- `c1.wheels = 6` does **not** change the class variable — it creates a new *instance* variable
  that shadows it for `c1` only.
- `self.x = x` in `__init__` → instance variable, unique per object.
- Method types:
  - **instance method** — first parameter `self`, called via an object
  - **`@classmethod`** — first parameter `cls`, callable via class *or* object
  - **`@staticmethod`** — no `self`/`cls`, callable via class *or* object
- A class cannot call an instance method without an object (`Student.avg()` → `TypeError`).
- **Inner class:** defined inside the parent (`Student.laptop.show()`).
- **Object as attribute (composition):** pass a `Laptop` object into `Student` — a "has-a" relation.
