# Variable Kinds & Multiple Inheritance

**OOP practice notes.** A quick recap of the three variable kinds (local,
instance, class) followed by **multiple inheritance** — one child inheriting from two or
more parent classes, with a real Saylani/AI-engineer example.

**What you will learn**
- Local vs instance vs class variables (scope & who owns them)
- Multiple inheritance syntax: `class Child(Parent1, Parent2):`
- Using `super()`/explicit parent calls to reach each parent's methods

### Kinds of variables in Python

- **instance variable** → one object
- **class variable** → shared by all objects
- **local variable** → inside a function/method only

```python
# defined inside another function/method
```

```python
# local variable: created inside show(), destroyed when the method finishes
class student:
  def show(self):
    name = "usman" # local variable
    print(name)
```

```python
s1 = student()
```

```python
s1.show()
```

```python
# instance variables belong to ONE object (self.name is unique per student)
class student:
  def __init__(self,name, age):
    self.name = name# instance variables
    self.age = age   # instance variable
```

```python
# class variable: husband_name is shared — read it via t1 or teacher directly
class teacher:
  husband_name = "Mustafa"  # class variables
  def __init__(self, subject, timing):
    self.subject  = subject
    self.timing = timing
  def show(self):
    age = 23
    return age
```

```python
t1 = teacher('math', "10pm-1am")
```

```python
t1.husband_name
```

```python

t1.timing
```

```python
t1.show()
```

## Multiple inheritance

A child class inherits from **2 or more** parent classes at the same time.

```python
# two independent parent classes, each with its own skill method
class Father:
  def father_skill(self):
    print("father  :     businnes maan" )

class Mother:
  def mother_skill(self):
    print("Mother  :    Cookiing")   # parent class

class child(Father, Mother):   # child class
  def child_skill(self):
    print('child  :   Coding')
```

```python
# the child object can call father_skill, mother_skill AND child_skill
c1 = child()
c1.father_skill()
c1.mother_skill()
c1.child_skill()
```

```python
# parent class 1: company + methods for the employment side
class Employe:  # parent class 1
  company  = "Saylani "  # class variables

  def employ_info(self):
    print("I ama an employee")

  def show_info(self):  # methods = functions defined inside a class
    print(f"company:   My company name is {self.company}")  # behaviour
```

```python
# parent class 2: AI-engineer methods
class AiEnginner:# parent class 2
  def ai_work(self):# methods
    print("I build Ai models")
  def train_model(self): # method
    print("Khud ka gpt banaya")
```

```python
# multiple inheritance: (Employe, AiEnginner) — child inherits from BOTH
class AIemployyy(Employe, AiEnginner):  # the child class inherits from BOTH parent classes
  def __init__(self, name, age):  # initializer / constructor
    self.name = name  # instance variable
    self.age = age
  def show_info(self): # # method
    print(f"Name    :    {self.name}")
    print(f"Age     :  {self.age}")
    print(f"company : {self.company}")
  def Employe_parent_show_info(self):  # new function that calls a parent class method and returns its value
    super().show_info()
```

```python
ai = AIemployyy("Usman", 20)
```

```python
# show_info() resolves through MRO — left parent wins on a name clash
ai.show_info()
```

```python
# calling a parent method explicitly by class name avoids the ambiguity
ai.Employe_parent_show_info()
```

## 🔑 Key Takeaways / Quick Revision

- **Local variable** → lives inside a function/method, gone after it returns.
- **Instance variable** (`self.x`) → belongs to ONE object, created in `__init__`.
- **Class variable** (`x = ...` in class body) → shared by ALL objects; readable via class or object.
- **Multiple inheritance:** `class Child(Parent1, Parent2):` gives the child BOTH parents' methods.
- Python resolves both parents **left to right** (MRO) if a method name collides.
- Call parents explicitly (`Employe.show_info(self)`) when you need a specific parent's version.
