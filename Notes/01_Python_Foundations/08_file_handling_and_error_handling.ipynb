**What you will learn**

## 1. Writing data to a file

- `open("read.txt", "w")` → **`"w"` mode creates the file if it does not exist and overwrites it** if it does.
- `with` is used so the file is **closed automatically** — even if the code inside raises an error.
- `encoding="utf-8"` ensures special characters (urdu/emojis) are stored correctly on every OS.

````python
try:
    # encoding='utf-8' keeps special characters safe across operating systems
    with open("read.txt", "w", encoding="utf-8") as f:
        f.write("hellow world!")

except FileNotFoundError:
    # raised when the folder/path in the filename does not exist
    print("File Not Found!")

except Exception:
    # catch-all: any other unexpected error is handled here
    print("Something went wrong")
````

## 2. Reading the file back

After writing, we open the same file in **read mode (`"r"`)** and print its contents to confirm
the data was actually saved.

- `"r"` → read only (default mode)
- `"a"` → append without deleting old content
- `"w"` → overwrite (previous data is lost!)

````python
# 'r' mode reads the existing content instead of overwriting it
with open("read.txt", "r", encoding="utf-8") as f:
    content = f.read()          # whole file comes back as one string

print(content)
````

## 3. How `try / except` works

```text
try:                       # code that MIGHT fail goes here
    risky_operation()
except FileNotFoundError:  # handle this specific error
    print("path not found")
except Exception:          # handle anything else
    print("something went wrong")
```

- Python **stops at the first `except` that matches** the error type.
- Putting `except Exception` last acts as a safety net so the program never crashes abruptly.
- Always catch the **narrowest** exception you expect; a bare catch-all can hide real bugs.

## 4. Practicing error handling

Here we intentionally read a file that does not exist to see `FileNotFoundError` being caught.

````python
try:
    with open("missing_file.txt", "r", encoding="utf-8") as f:
        print(f.read())

except FileNotFoundError:
    # the program continues instead of crashing — that is the whole point of try/except
    print("Handled gracefully: the file does not exist yet.")
````

## 🔑 Key Takeaways / Quick Revision

- `with open(path, mode, encoding="utf-8") as f:` → safest way to open a file (auto-closes).
- Modes: `"r"` read • `"w"` overwrite • `"a"` append.
- `f.write(text)` writes, `f.read()` reads whole file, `f.readlines()` reads line by line.
- `try:` holds risky code • `except SomeError:` handles that specific error • `except Exception:` is the last-resort net.
- Always prefer **specific** exceptions (`FileNotFoundError`, `ValueError`, …) over a bare `except:`.
