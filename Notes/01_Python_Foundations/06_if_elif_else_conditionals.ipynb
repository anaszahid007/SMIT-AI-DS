## if statement — running code only when a condition is True

```text
if (condition) -> True:
    ------------------
    ------------------
```

- Python checks the condition and enters the indented block **only** if it is `True`.
- The condition is any expression that evaluates to a bool (`age > 18`, `name == "Usman"`).
- `else:` provides the alternative block when the condition is `False`.

```text
if (condition) -> True:
    ------------------
    ------------------
```

````python
if True:
  print("hz")
````

````python
if True:
  print("hz")
````

````python
# if/else: exactly one branch runs depending on the comparison result
name = input("what is your name ?")
if name == 'Usman':
  print("the person is authentic")
else:
  print("The person is not Authentic")
````

### Program: discount using if / else

- Take the bill amount, then branch:
  - `BillAmount >= 20000` → 20% discount
  - otherwise → print "cannot avail the discount offer"
- Demonstrates a complete `if / else` pair where **exactly one** branch runs.

````python
# first branch: eligible for 20% discount when the bill is 20000 or more
BillAmount = float(input("Kindly Enter your Bill Amount :"))
if BillAmount >= 20000.0 :
  discount = BillAmount * 0.20
  print(f"Your discount in RS:{discount} on the purchase of RS:{BillAmount}")
else:
  print(f"you have Purchased amount is {BillAmount}, so you cant avail the Discount Offer")
````

### elif — multiple mutually exclusive branches

- `elif` adds more conditions **after** `if`; Python checks them top-down and runs
  the **first** branch whose condition is `True` — the rest are skipped.
- Examples below: discount slabs by amount, eligibility by age with `and`,
  and comparing two names with `if / elif / else`.

````python
# elif chain: Python stops at the FIRST slab that matches
BillAmount = float(input("Kindly Enter your Bill Amount :"))
if BillAmount >= 20000.0 :
  discount = BillAmount * 0.20
  print(f"Your discount in RS:{discount} on the purchase of RS:{BillAmount}")

else:
  print(f"you have Purchased amount is {BillAmount}, so you cant avail the Discount Offer")
````

````python
# two conditions combined with and -> covers the 12..17 age range
age = int(input("what is yout age: "))
if age >= 18:
  print("business can be done")
elif age < 18 and age >= 12:
  print("you need to learn how to run business")
else:
  print("just do what you want")
````

````python
# total -> percentage -> grade: conditionals chained in sequence
python = int(input("Enter python marks"))
numpy = int(input("Enter numpy marks"))
pandas  = int(input("Enter pandas marks"))
stats = int(input("Enter stats marks"))
mL = int(input("Enter mL marks"))
total = python + numpy + pandas + stats + mL
percentage  = total / 500 * 100
if percentage>=90:
    grade  = "A+"
elif percentage>=80:
    grade = "A"
elif percentage>=70:
    grade = "B"
elif percentage>=60:
    grade = "C"
elif percentage>=50:
    grade = "D"
else:
    grade = "F"
print("Total: ", total)
print("Percentage: ", percentage)
print("Grade: ", grade)
````

````python
items = {'laptop':50000, 'mobile':60000, 'smartwatch':10000,'earbuds':5000}
````

````python
# in + get() never crash on an unknown key — the safe pattern
item_name = input("Enter the item you want to buy")
if item_name in items:
    print(f"The price of {item_name} is: {items.get(item_name)}")
else:
    print(f"The {item_name} is not available at our store")
````

## 🔑 Key Takeaways / Quick Revision

- Syntax: `if condition:` → block runs only when the condition is `True` (indentation matters).
- `else:` handles the "nothing matched" case — it takes no condition.
- `elif condition:` adds extra branches; Python runs **only the first matching branch**.
- Conditions combine values: `age >= 18`, `name == 'Usman'`, `price > 200`, `x and y`.
- Read numbers with `float(input(...))` / `int(input(...))` so comparisons work.
- Check a dict key safely: `if key in items:` or `items.get(key, default)` — direct
  `items[key]` raises `KeyError` for unknown keys.
- Practice programs covered: bill discount, temperature check, tomato price slabs,
  subject marks → percentage → grade, item price lookup.
