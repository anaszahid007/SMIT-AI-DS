## Boolean / Comparison Operators

- `==` equal to • `!=` not equal to • `>` greater than • `<` less than
- `>=` greater than or equal • `<=` less than or equal
- Every comparison expression evaluates to a **bool**: `True` or `False`.
- Careful: `10/3 == 3.3` is `False` because `10/3` is `3.333...` — decimal results can differ.

```python
# store the numbers in variables first so the comparisons below stay readable
asif = 23
arham = 10
taham = 23
```

```python
arham > asif
```

```python
arham < asif
```

```python
taham == asif
```

```python
taham>=asif
```

```python
ayub = 16
```

```python
ayub <> arham
```

> ⚠️ `<>` is the **Python 2** not-equal operator. Python 3 raises `SyntaxError` —
> always write `ayub != arham` instead.

## Logical Operators

- `and` → True only when **both** expressions are True
- `or` → True when **at least one** expression is True
- `not` → flips True to False and False to True

| Exp1 | Exp2 | `Exp1 and Exp2` | `Exp1 or Exp2` |
|---|---|---|---|
| True | True | True | True |
| True | False | False | True |
| False | True | False | True |
| False | False | False | False |

```python
# or needs only ONE True expression to return True
12>3 or 13>1 or 15<10
```

```python
# and needs ALL expressions to be True
12>3 and 3>1 and 15<10
```

```python
x = -100
from math import sqrt
x > 0 or sqrt(x)
```

> ⚠️ **Short-circuit demo:** with `x = -100`, `x > 0` is `False`, so Python *does*
> evaluate `sqrt(-100)` → `ValueError: math domain error`. Use `x > 0 and sqrt(x)` when the
> right side is only safe if the left side is True.

## 🔑 Key Takeaways / Quick Revision

- `print(value)` displays output; `type(value)` tells its data type.
- Data types: `str` (quotes) • `int` • `float` • `bool`.
- Arithmetic: `/` divides (float) • `//` floor-divides • `%` gives remainder • `**` raises to power.
- Comparison (`==  !=  >  <  >=  <=`) always returns `True` or `False`.
- `and` → True only if **both** sides True • `or` → True if **any** side True • `not` flips the value.
- Empty values are *falsy*: `0`, `""`, `[]`, `{}`, `False`; everything else is usually *truthy*.
- `<>` (not-equal) was Python 2 — Python 3 raises `SyntaxError`; always use `!=`.
- `and`/`or` **short-circuit**: Python skips the right side when the answer is already decided.
