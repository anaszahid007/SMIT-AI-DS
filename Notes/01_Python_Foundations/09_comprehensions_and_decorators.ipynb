## List comprehension — build a list in one line

- Provides a concise syntax to create a new list **from an existing iterable**:
  `[expression for item in iterable]`
- With a filter: `[expression for item in iterable if condition]`
- With if/else: `[value_if_true if condition else value_if_false for item in iterable]`
- Compare every example with the equivalent `for` loop written above it — same result, less code.

```python
lst = [2,4,6,5,8,9,1,45,2]
```

```python
squared_lst = []
for n in lst:
    sq = n**2
    squared_lst.append(sq)
squared_lst
```

```python
# comprehension version of the same loop above — one line instead of four
[n**2 for n in lst]
```

```python
lst = [2,4,6,5,8,9,1,45,2]
```

```python
new_lst = []
for n in lst:
    if n%2==0:
        sq = n**2
        new_lst.append(sq)
    else:
        cub = n**3
        new_lst.append(cub)
new_lst
```

```python
# if/else inside a comprehension: square even numbers, cube the odd ones
[n**2 if n%2==0 else n**3 for n in lst]
```

```python
percentage  = float(input("Percentage: "))
if percentage>=90:
    grade = "A+"
elif percentage>=80:
    grade='A'
elif percentage>=70:
    grade='B'
elif percentage>=60:
    grade='C'
elif percentage>=50:
    grade='D'
else:
    grade="F"
print(grade)
```

## Dictionary comprehension — build a dict in one line

- Syntax: `{key_expression: value_expression for item in iterable}`
- Example: `{n: n**2 for n in lst}` maps every number to its square.
- Example: `{c: len(c) for c in country}` maps each country name to its length.

```python
# dict comprehension: {key: value for ...} builds the whole mapping at once
{n:n**2 for n in lst}
```

```python
{c:len(c) for c in country}
```

## Inner functions (local functions)

- A function defined **inside** another function exists only while the parent runs.
- Calling it from outside raises `NameError` — it is out of scope after `parent()` finishes.
- Inner functions are the "wrapper" half of every decorator.

```python
# inner functions live inside the parent — out of scope afterwards
def parent():
    print("Printing from parent()")

    def first_child(): # local function
        print("Printing from first_child()")

    def second_child(): # local function
        print("Printing from second_child()")

    second_child() # local function calling inside function
    first_child()  # local function calling inside function
```

```python
parent() # Usual call to a function
```

```python
first_child() # result in an error
```

> ⚠️ Expected error: `first_child()` was defined **inside** `parent()` — after `parent()`
> finishes, the inner function is out of scope → `NameError`.

## Returning functions from functions

- `parent()` returns the **function object** `first_child` itself (not its result).
- Storing it (`first = parent(1)`) and later calling `first()` proves functions are objects.
- Each call to `parent()` builds a fresh inner function — that is what decorators exploit.

```python
def parent(num):
    def first_child():
        return "Hi, I'm Elias"

    def second_child():
        return "Call me Ester"

    if num == 1:
        return first_child  # returning a reference
    else:
        return second_child # returning a reference
```

```python
parent(4)
```

```python
parent(2)
```

## A simple decorator

- `decorator(func)` receives a function, defines a `wrapper()` inside, and returns it.
- `say_whee = decorator(say_whee)` replaces the original function with the wrapper —
  so calling `say_whee()` now runs the extra `print(...)` lines **around** the original code.

```python
# decorator = takes a function, defines a wrapper, returns the wrapper
# this is a decorator function. It accepts a function as a parameter
def decorator(func):
    def wrapper(): # the inner function can be named anything
        print("Something is happening before the function is called.")
        func()
        print("Something is happening after the function is called.")
    return wrapper
```

```python
def say_whee():
    print("Whee!")
```

```python
# this assignment line is what @decorator does automatically (see later)
say_whee = decorator(say_whee)
```

```python
say_whee
```

```python
say_whee()
```

##### However, wrapper() has a reference to the original say_whee() as func, and it calls that function between the two calls to print().

##### Put simply, a decorator wraps a function, modifying its behavior.

##### To see both halves of this in motion, step through the widget below. The first tab shows what happens when Python runs the decoration ##### line, say_whee = decorator(say_whee), and the second shows what happens when you then call say_whee():

```python
from datetime import datetime

def not_during_the_night(func):
    def wrapper():
        # class wrote "713 <=" by mistake — must be 7 <= (7 AM .. 9 PM)
        if 7 <= datetime.now().hour < 22:
            func()
        else:
            print("Too late, sleeping instead")
    return wrapper
```

```python
def say_whee():
    print("Whee!")
```

```python
say_whee = not_during_the_night(say_whee)
```

```python
say_whee()
```

## Applying a decorator — the final form

- The wrapper adds work before **and** after `func()` runs.
- Next: the `@decorator` syntax, which is Python's shortcut for the assignment line above.

```python
def decorator(func):
    def wrapper():
        print("Something is happening before the function is called.")
        func()
        print("Something is happening after the function is called.")
    return wrapper
```

## The @ syntax explained

- `@decorator` written above a function is **exactly** the same as
  `say_whee = decorator(say_whee)` — Python does the replacement for you.
- `@do_twice` runs the wrapped function **two times**; wrappers with `(*args, **kwargs)`
  can decorate functions that take arguments too (and must `return` their result).

```python
# @decorator is the shortcut for say_whee = decorator(say_whee)
@decorator
def say_whee():
    print("Whee!")
```

```python
say_whee()
```

```python
def do_twice(func):
    def wrapper_do_twice():
        func()
        func()
    return wrapper_do_twice
```

```python
@do_twice
def say_whee():
    print("Whee!")
```

```python
say_whee()
```

```python
@do_twice
def greet(name):
    print(f"Hello {name}")
```

```python
greet('nasir')
```

```python
# wrapper with *args/**kwargs can decorate functions that take arguments
def do_twice(func):
    def wrapper_do_twice(*args, **kwargs):
        func(*args, **kwargs)
        func(*args, **kwargs)
    return wrapper_do_twice
```

```python
@do_twice
def greet(name):
    print(f"Hello {name}")
```

```python
greet(name='nasir')
```

```python
greet('asad')
```

```python
@do_twice
def return_greeting(name):
    print("Creating greeting")
    return f"Hi {name}"
```

```python
return_greeting('nasir')
```

```python
hi_adam = return_greeting("Adam")
```

## Iterators — value by value on demand

- An iterator is an object holding a **countable** number of values, produced one at a time.
- `iter(collection)` creates an iterator • `next(iterator)` fetches the next value.
- When values run out, `next()` raises **`StopIteration`**.

An iterator is an object that contains a countable number of values.

An iterator is an object that can be iterated upon, meaning that you can traverse through all the values.

Technically, in Python, an iterator is an object which implements the iterator protocol, which consist of the methods __iter__() and __next__().1mr

## Iterable vs Iterator

- **Iterable:** a container you can loop over — list, tuple, dict, set, string.
- **Iterator:** the object that remembers its position during iteration (`iter(list)`).
- A `for` loop calls `iter()` automatically and keeps calling `next()` until `StopIteration`.

Lists, tuples, dictionaries, and sets are all iterable objects. They are iterable containers which you can get an iterator from.

All these objects have a iter() method which is used to get an iterator:

```python
# iter() turns an iterable into an iterator with a remember-position
myList = ["Honda", "Suzuki", "Cherry"]
my_list_iterator = iter(myList)
```

## Generators — functions that pause and resume

- A generator function contains `yield`; calling it returns a **generator object**.
- Each `next()` runs the body until the next `yield`, then **pauses** exactly there.
- The function keeps its local state between pauses — like a saved bookmark.

Generators are functions that can pause and resume their execution.

When a generator function is called, it returns a generator object, which is an iterator.

The code inside the function is not executed yet, it is only compiled. 

The function only executes when you iterate over the generator.

## Why generators?

- **Memory efficient:** huge or infinite sequences without loading everything into RAM.
- **No big list built:** values are produced only when requested (lazy evaluation).

## yield vs return

- `return` ends the function **immediately** and hands back a single value.
- `yield` hands back one value and **pauses** — the next `next()` continues from that line.
- A generator can `yield` many times; after the last one, it raises `StopIteration`.

```python
def my_generator():
  yield 1
  yield 2
  yield 3

for value in my_generator():
  print(value)
```

```python
# yield pauses the generator right there and resumes on the next next()
def count_up_to(n):
  count = 1
  while count <= n:
    yield count
    count += 1
```

```python
def count_up_to(n):
  count = 1
  while count <= n:
    yield count
    count += 1
```

```python
for num in count_up_to(5):
  print(num)
```

```python
def fun(max):
    cnt = 1
    while cnt <= max:
        yield cnt
        cnt += 1

ctr = fun(5)
for n in ctr:
    print(n)
```

- `(x*x for x in range(1, 6))` → parentheses create a **generator**: values appear on demand
  while looping.

## 🔑 Key Takeaways / Quick Revision

- List comprehension: `[expr for x in iterable if condition]` — replaces a whole `for` loop.
- Conditional expression: `[expr1 if cond else expr2 for x in iterable]`.
- Dict comprehension: `{key: value for x in iterable}` → e.g. `{c: len(c) for c in countries}`.
- Functions are **objects**: pass them as arguments, return them, store them in variables.
- **Inner functions** exist only inside their parent — calling them outside raises `NameError`.
- Decorator = a function that takes a function and returns a **wrapper** that runs extra code
  before/after it; `@decorator` is short for `fn = decorator(fn)`.
- Wrappers that accept arguments need `(*args, **kwargs)` and must **return** the result.
- `iter(obj)` creates an iterator, `next(it)` fetches one value — exhausted → `StopIteration`.
- Generator functions use `yield`: they **pause** between values and keep almost no memory;
  `return` ends a function completely.
- Generator expression `(x*x for x in range(6))` builds no list — values are produced on demand.
