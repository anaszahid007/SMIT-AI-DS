## Pandas — Python's data analysis library

- pandas is a **library** built for working with real-world (tabular) data.
- It deals with all kinds of data: numbers, text, dates, missing values...
- **Under the hood it runs on NumPy**, so operations are fast.
- Two core structures:
  > **Pandas Series** → one column (1-D) • **Pandas DataFrame** → a whole table (2-D)

```python
# pd.Series(data, index=...) — 1-D labelled array, the basic pandas object
import pandas as pd

s1 = pd.Series([1,2,3,4,5,6,7])
shope_products = pd.Series(['computer', 'laptop', 'mobile', 'watch', 'keyboard', 'mouse'])
print(f'length of the s1 series: {len(s1)}')
print(f'length of the shop products: {len(shope_products)}')

print(s1)
print(shope_products)
```

**Output of every Series shows three things:** `index` • `value` • `dtype` (data type).

```python
# slicing works on a Series just like on a list
shope_products[1:3]
```

```python
# custom index lets you fetch values by LABEL instead of position
# changing their index

fruits = pd.Series(data=[1,2,3,4], index=['apple', 'mango','banana','orange'])

print(fruits['mango': 'banana'])
print(fruits)
```

## DataFrame — from a dictionary to a full results table

- Each dictionary key becomes a **column**; its list becomes the column's values.
- Then we compute new columns (`total`, `percentage`), assign `grade`/`status` with list
  comprehensions, and rank the table with `sort_values()`.

```python
# DataFrame from a dict of equal-length lists -> one column per key
def split_line():
    print('\n')
    print('----------------------------------------------------------')

data = {
    "name": ['Anas', 'Salaudeen', 'Saad', 'Kashif', 'Hassan'],
    "roll_number": [123, 234, 245, 345, 347],
    "python_marks": [82, 71, 94, 83, 75],
    "pandas_marks": [85, 73, 91, 82, 79],
    "math": [72, 83, 93, 62, 82]
}

df = pd.DataFrame(data)

# new column = sum of the three subject columns
df['total'] = df.python_marks + df.pandas_marks + df.math
# percentage of 300 total marks
df['percentage'] = df.total / 300 * 100

# grade column built with a chained if/elif/else comprehension
df['grade'] = [
    'A' if i >= 90
    else 'B' if i >= 80
    else 'C' if i >= 60
    else 'D' if i >= 50 else 'F'
    for i in df.percentage]

# pass/fail derived from the grade column
df['status'] = ['Fail' if i == 'F' else 'Pass' for i in df.grade]

# print(df.info())
# split_line()

# rank the whole table by total marks (highest first)
print(df.sort_values(ascending=False, by="total"))
# split_line()

# select only two columns: df[['name', 'roll_number']]
```

## 🔑 Key Takeaways / Quick Revision

- `import pandas as pd` — pandas handles all kinds of tabular data; NumPy runs underneath.
- Two structures: **Series** (1-D, index + value + dtype) • **DataFrame** (rows × columns).
- `pd.Series([1,2,3])` → default numeric index; `pd.Series(data=[...], index=[...])` → labels.
- Label slicing works on a custom index: `fruits['mango':'banana']`.
- `pd.DataFrame(dict_of_lists)` → each dict key becomes a column (equal lengths required).
- Derived columns: `df['total'] = df.a + df.b + df.c` • `df['percentage'] = df.total / 300 * 100`
- Grades/status via list comprehension over `df.percentage`.
- `df.sort_values(by='total', ascending=False)` ranks rows • `df[['col1','col2']]` selects columns
  • `df.info()` shows column types and missing values.
