**Pandas class 2 notes.** Working with a real Excel file (`employee.xlsx`):

**What you will learn**

## 1. Loading the Excel file & first look

- `pd.read_excel('data/employee.xlsx')` reads the sheet from the `data/` folder.
- Then: `sample()` (random rows), `iloc[0:5]` (positional slice), `head()` (first 5 rows).

```python
# read_excel() loads the sheet; the file lives in the data/ folder
import pandas as pd

# the Excel file lives in the data/ folder next to this notebook
data = pd.read_excel('data/employee.xlsx')
print(data)
```

```python
# sample() returns random rows — useful to spot-check messy data
print(data.sample(3))
```

```python
# iloc uses POSITIONS: rows 0..4 (end excluded, like Python slices)

print(data.iloc[0:5])
```

```python
# head() = quick look at the first five rows
print(data.head())
```

## 2. Selecting rows & columns — loc vs iloc

- `iloc[0:5]` → integer positions, end **excluded**.
- `loc[2:8, 'Designation':'Age']` → labels, end **included** (row 2 through 8).
- `loc[[1,3,4,5,6], ['Designation','Age']]` → exact row ids and column names.

```python
# loc uses LABELS: rows 2..8 AND columns Designation..Age (end INCLUDED)
print(data.loc[2:8, 'Designation':'Age'])
```

```python
# exact lists in loc pick specific row ids and column names
print(data.loc[[1,3,4,5,6],['Designation', 'Age']])
```

## 3. Boolean filtering — keep only the rows you want

- `df[df.Designation == "Manager"]` → the mask `df.Designation == "Manager"` is a True/False
  series; pandas keeps only the `True` rows.
- Combine conditions with `&` (and) / `|` (or) — each condition needs `( )` around it.

```python
# boolean mask: only rows where Designation equals Manager survive
print(data[data.Designation=="Manager"])
```

```python
print(data[data.Designation=="Accountant"])
```

```python
print(data[data.Age>=30])
```

```python
print(data[data.Salary>=100000])
```

```python
# & combines two masks — note the parentheses around EACH condition

print(data[(data.Designation == 'Engineer') & (data.Salary <= 100000)])
```

```python

print(data[(data.Designation == 'Accountant') | (data.Age >45)])
```

## 4. Aggregations & comparing against averages

- `.max()` / `.mean()` summarise a whole column into one number.
- `df[df.Salary < df.Salary.mean()]` filters rows against their own column average.

```python
# max() collapses the Salary column to its highest value
max_sal = data.Salary.max()
print(max_sal)
```

```python
# mean() gives the average — used as a threshold in the next cells
avg  = data.Salary.mean()
print(int(avg))
```

```python
data[data.Salary < data.Salary.mean()]
```

```python
data[data.Age > data.Age.mean()]
```

## 5. describe() — statistics of the whole table

- `df.describe(include='all')` gives count, mean, min, max and quartiles for **every** column —
  the standard first step of any analysis.

```python
# describe(include="all") = one-shot statistical summary of all columns
data.describe(include='all')
```

## Understanding data mathematically — types of analysis

- **Univariate analysis** → one variable at a time (e.g. Salary alone): `max()`, `mean()`, `describe()`
- **Bivariate analysis** → two columns together (e.g. Age vs Salary): comparisons, correlation
- **Multivariate analysis** → three or more columns at once (grouping, correlation matrix)
