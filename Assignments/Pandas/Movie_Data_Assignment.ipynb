### **Question 26**
(Visualization) Dataset mein 'Certificate' wise average 'Gross' kamai ka ek Bar Plot banayein.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd


data['Gross_Cleaned'] = data['Gross'].astype(str).str.replace(',', '', regex=False)
data['Gross_Cleaned'] = pd.to_numeric(data['Gross_Cleaned'], errors='coerce')


avg_gross_by_cert = (
    data.dropna(subset=['Certificate', 'Gross_Cleaned'])
    .groupby('Certificate')['Gross_Cleaned']
    .mean()
    .sort_values(ascending=False)
)


sns.set_theme(style="whitegrid")
plt.figure(figsize=(12, 6))


ax = sns.barplot(
    x=avg_gross_by_cert.values, 
    y=avg_gross_by_cert.index, 
    palette='magma'
)

for container in ax.containers:
    labels = [f"${val*1e-6:.1f}M" if val >= 1e6 else f"${val:,.0f}" for val in container.datavalues]
    ax.bar_label(container, labels=labels, padding=5, fontweight='bold', fontsize=10)

def millions_formatter(x, pos):
    return f"${x*1e-6:.0f}M" if x >= 1e6 else f"${x:,.0f}"
plt.gca().xaxis.set_major_formatter(plt.FuncFormatter(millions_formatter))

plt.title('Average Gross Earnings by Movie Certificate Type', fontsize=15, fontweight='bold', pad=20)
plt.xlabel('Average Gross Revenue (in Millions)', fontsize=12, fontweight='bold')
plt.ylabel('Certificate', fontsize=12, fontweight='bold')

sns.despine(left=True, bottom=True)
plt.tight_layout()

plt.show()
```

### **Question 27**

### **Question 36**
(Visualization) 'Drama' aur 'Action' genres ki movies ke 'Gross' revenue ko compare karne ke liye ek Grouped Box Plot ya Violin Plot banayein.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

data['Gross_Cleaned'] = data['Gross'].astype(str).str.replace(',', '', regex=False)
data['Gross_Cleaned'] = pd.to_numeric(data['Gross_Cleaned'], errors='coerce')

genre_str = data['Genre'].fillna('').astype(str)
action_movies = data[genre_str.str.contains('Action', case=False)].copy()
drama_movies = data[genre_str.str.contains('Drama', case=False)].copy()

action_movies['Target_Genre'] = 'Action'
drama_movies['Target_Genre'] = 'Drama'

comparison_df = pd.concat([action_movies, drama_movies]).dropna(subset=['Gross_Cleaned'])

sns.set_theme(style="whitegrid")
plt.figure(figsize=(10, 6))

sns.violinplot(
    data=comparison_df, 
    x='Target_Genre', 
    y='Gross_Cleaned', 
    palette='muted',
    inner='quart' 
)

def millions_formatter(x, pos):
    return f"${x*1e-6:.0f}M" if x >= 1e6 else f"${x:,.0f}"
plt.gca().yaxis.set_major_formatter(plt.FuncFormatter(millions_formatter))

plt.title('Gross Revenue Comparison: Action vs. Drama Movies', fontsize=15, fontweight='bold', pad=20)
plt.xlabel('Genre', fontsize=12, fontweight='bold')
plt.ylabel('Gross Revenue (in Millions)', fontsize=12, fontweight='bold')

sns.despine()
plt.tight_layout()

plt.show()
```

### **Question 40**
(Visualization) Top 5 highest rated movies aur unke 'Meta_score' vs 'IMDB_Rating' (scaled to 100) ka ek Side-by-Side Comparison Bar Chart banayein.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

data['IMDB_Rating_Clean'] = pd.to_numeric(data['IMDB_Rating'], errors='coerce')
data['Meta_score_Clean'] = pd.to_numeric(data['Meta_score'], errors='coerce')
data['IMDB_100_Scale'] = data['IMDB_Rating_Clean'] * 10

top_5_movies = data.sort_values(by='IMDB_Rating_Clean', ascending=False).head(5)

comparison_df = top_5_movies[['Series_Title', 'IMDB_100_Scale', 'Meta_score_Clean']]

melted_df = pd.melt(
    comparison_df, 
    id_vars=['Series_Title'], 
    value_vars=['IMDB_100_Scale', 'Meta_score_Clean'],
    var_name='Score_Type', 
    value_name='Rating_Value'
)

melted_df['Score_Type'] = melted_df['Score_Type'].replace({
    'IMDB_100_Scale': 'IMDb Rating (x10)',
    'Meta_score_Clean': 'Meta Score'
})

sns.set_theme(style="whitegrid")
plt.figure(figsize=(12, 6))

ax = sns.barplot(
    data=melted_df, 
    x='Series_Title', 
    y='Rating_Value', 
    hue='Score_Type', 
    palette='Set2'
)

for container in ax.containers:
    ax.bar_label(container, padding=3, fmt='%.0f', fontweight='bold', fontsize=10)

plt.title('Side-by-Side Comparison: IMDb Rating vs Meta Score (Top 5 Movies)', fontsize=15, fontweight='bold', pad=20)
plt.xlabel('Movie Title', fontsize=12, fontweight='bold')
plt.ylabel('Score (out of 100)', fontsize=12, fontweight='bold')
plt.ylim(0, 110)
plt.xticks(rotation=15, ha='right')

sns.despine()
plt.tight_layout()

plt.show()
```
