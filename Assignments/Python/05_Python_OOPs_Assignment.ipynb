**Q1:** Aik `Student` class banayein jismein `name` aur `roll_number` private variables (`__`) hon. In variables ko access aur modify karne ke liye getter aur setter methods likhein.

**Q2:** Aik `BankAccount` class banayein jismein `__balance` private variable ho. Aik `deposit()` aur aik `withdraw()` method banayein, check karein ke withdraw karte waqt balance zero se kam na ho.

```python
# Write your code here for Q2
class BankAccount:
    __balance = 0

    def deposit(self, amount: int):
        self.__balance = amount

    def withdraw(self, withdraw_amount):
        if self.__balance >= withdraw_amount:
            self.__balance -= withdraw_amount
            print(f"WithDraw Balance: {withdraw_amount}\nRemaining Balance: {self.__balance}")
        else:
            print('balance is less than you withdraw amount.')

b1 = BankAccount()
b1.deposit(1000)
b1.withdraw(500)
```

```python
# Write your code here for Q4

### Protected and Private Variables in Python
# 1. Protected variables are those that should not be accessed outside of the class and just accessed by its own and child classes. 
    # - They are indicated by a single underscore prefix (e.g., _variable). 

# 2. Private variables are those that should not be accessed outside of the class even in the child classes. 
    # - They are indicated by a double underscore prefix (e.g., __variable).

class Employee:
    def __init__(self, name, salary, department):
        self._name = name # Protected variable
        self.__salary = salary # Private variable
        self.__department = department # Private variable
    
    def display_info_employee(self): # Method through Error cuz it's trying to access private variables of parent class which is not allowed in Python.
            print(f"Manager Name: {self._name}, Salary: {self.__salary}, Department: {self.__department}")
    

class Manager(Employee):
    def __init__(self, name, salary, department):
        super().__init__(name, salary, department)

    def display_info_manager(self): # Method through Error cuz it's trying to access private variables of parent class which is not allowed in Python.
        print(f"Manager Name: {self._name}, Salary: {self.__salary}, Department: {self.__department}")

m1 = Manager('Anas', 1000, 'IT')
# m1.display_info_manager() # This will throw an error because __salary and __department are private variables in the Employee class and cannot be accessed directly in the Manager class.
m1.display_info_employee() # This will work because _name is a protected variable and can be accessed in the child class. However, __salary and __department are private variables and can be accessed directly in the Employee class.
```

**Q10:** Hierarchical Inheritance kya hoti hai? Aik parent class `Shape` banayein aur us se do alag child classes `Circle` aur `Square` banakar samjhayein.

```python
# Write your code here for Q10

# Hierarchical Inheritence os 'pattern' ko kehte hai jahan ek 'parent' class hota hai or baki 'child class', 'parent class' ko inherit karte hai esko Hierarchical Inheritence kehte hai
# for example ap code ko dek skte hai jesa k mene hierarchical inheritence use kya hai.

class Shape:
    def __init__(self, call_from):
        print('Shape class called from: ', call_from, 'class')

class Circle(Shape):
    def __init__(self):
        super().__init__("Circle")

class Square(Shape):
    def __init__(self):
        super().__init__("Square")


c1 = Circle()
s1 = Square()
```

**Q13:** Operator Overloading kya hoti hai? Do objects ko `+` operator ke zariye add karne ke liye `__add__` magic method ka use kaise karte hain? Code likhein.

```python
# Write your code here for Q13
class Distance():
    def __init__(self, kilometers):
        self.kilometers = kilometers

    def __add__(self, other):
        return Distance(self.kilometers + other.kilometers)

d1 = Distance(10)
d2 = Distance(20)
d3 = d1 + d2

print(f"Total Distance: {d3.kilometers} km")
```

**Q15:** Kya Python default roop se Method Overloading (same method name with different arguments) support karta hai? Agar nahi, toh hum same name ke method ko multiple arguments ke sath kaise handle karte hain? Code likhein.

```python
# Write your code here for Q15


# Python default roop se method overloading support nahi karta. 
# Isay handle karne ke liye hum *args (Variable-length arguments) ka use karte hain, 
# jo method ko multiple arguments accept karne ki flexibility deta hai.


class Calculator():
    def add(self,*args):
        return sum(args)

c = Calculator()

print(c.add(10, 20,23,454,2323,4,233,23))
print(c.add(3, 4))
```

**Q18:** Aik abstract class `Car` banayein jisme aik concrete (normal) method `start_engine()` ho aur aik abstract method `drive()` ho. Child class bana kar dono ko test karein.

```python
# Write your code here for Q18
from abc import ABC, abstractmethod

class Car(ABC):
    @abstractmethod
    def drive():
        pass

class HiluxTruck(Car):
    def start_engine(self):
        print('engine started')

    def drive(self):
        print("driving car")

h = HiluxTruck()
h.start_engine()
h.drive()
```

**Q20:** Aik real-world scenario (jaise Payment System) design karein jahan `Payment` aik abstract class ho aur `CreditCardPayment` aur `UPIPayment` uski subclasses hon.

```python
# Write your code here for Q20
from abc import ABC, abstractmethod
class Payment(ABC):

    @abstractmethod
    def check_balance(self):
        pass

    @abstractmethod
    def deposit(self, amount):
        pass

    @abstractmethod
    def CashWithDraw(self, amount):
        pass

    
class CreditCardPayment(Payment):
    def __init__(self):
        self.__balance = 0

    def check_balance(self):
        return self.__balance

    def deposit(self, amount):
        self.__balance += amount
        print(f"Rs. {amount} deposited via Credit Card.")
        print(f"Total balance is: Rs. {self.__balance}\n")

    def CashWithDraw(self, amount):
        if self.__balance >= amount:
            self.__balance -= amount
            print(f"Rs. {amount} withdrawn via Credit Card.")
            print(f"Remaining balance is: Rs. {self.__balance}\n")
        else:
            print("Transaction Failed: Amount is greater than the balance.\n")


class UPIPayment(Payment):
    def __init__(self):
            self.__balance = 0
    
    def check_balance(self):
        return self.__balance

    def deposit(self, amount):
        self.__balance += amount
        print(f"Rs. {amount} deposited via UPI Mobile Wallet.")
        print(f"Total balance is: Rs. {self.__balance}\n")

    def CashWithDraw(self, amount):
        if self.__balance >= amount:
            self.__balance -= amount
            print(f"Rs. {amount} withdrawn via UPI.")
            print(f"Remaining balance is: Rs. {self.__balance}\n")
        else:
            print("Transaction Failed: Insufficient UPI Balance.\n")


# Testing the System
print("--- CREDIT CARD TRANSACTION ---")
card = CreditCardPayment()
card.deposit(4000)
card.CashWithDraw(3000)
card.CashWithDraw(6000)

print("--- UPI TRANSACTION ---")
upi = UPIPayment()
upi.deposit(500)
upi.CashWithDraw(1200)
```
