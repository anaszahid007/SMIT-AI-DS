### Q5

```python
# Q5 solution code here
with open('04_handle_files/numbers.txt', 'r') as fs:
    numbers = fs.readlines()
    sum = 0
    for i in numbers:
        sum += int(i)
    print(sum)
```

### Q6 
Ek Python program likhein jo kisi file mein total lines count karke bataye.

```python
# Q6 solution code here

def count(numbers):
        count = 0
        for _ in numbers:
            count += 1
        return count

with open('04_handle_files/numbers.txt', 'r') as fs:
    numbers = fs.readlines()
    print(count(numbers))
```

### Q17
Ek `try-except` block likhein jo Multiple Exceptions catch kar sake (`ZeroDivisionError` aur `ValueError`).

```python
# Q17 solution code here
try:
    nm = int(input('enter a number'))

    print(10/0)
    print(nm)
    
except ValueError:
    print('Error: Not a number')

except ZeroDivisionError:
    print("Error: Can't be divide with 0")
    
except Exception:
    print("Error: something went wrong")
```

### Q18

### Q19 
`raise` keyword ka use karke age validation ka function banayein. Agar age 18 se kam ho toh custom `ValueError("Age must be 18 or older")` raise karein.

```python
# Q19 solution code here
try:
    age = int(input('Enter your age'))
    if age <= 18:
        raise ValueError('Age must be 18 or older')
    elif age > 100:
        raise ValueError("Age can't be older than 100")
    
    print('your: ', age)

except ValueError as e:
    print(e)

except Exception:
    print('Soemthing went Wrong.!')
```
